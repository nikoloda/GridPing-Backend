```julia
# This file takes in a meter reading/ grid state timestamp, then uploads the data to the DB specified by the user

# Initialize global/ configurable variables (Anything which only needs to be run once)

# Dataframes adds ability to visualize nicer
using DataFrames

# For mean calculation
using Statistics

# Grid Optimizer
using Ipopt

# Select which DB(s) to populate
include("../src/DB_AWS_PostgreSQL.jl")
include("../src/DB_SQLite.jl")

const USE_AWS_POSTGRES = true
const USE_LOCAL_SQLITE = false
    
    
# The bus id where the smart meter is located
meter_bus_id = 6

# Pretend we recieve an update from the utility, triggering the following code
# This code will contain the matpower case file at the most recent time
using PowerModels

# Open ideal "MATPOWER grid state case"
sys = PowerModels.parse_file("../cases/transmission/case14.m")

# received_sys = PowerModels.parse_file("../cases/transmission/case14.m")

# Ideal time stamp (Will be the earliest time in our DB)
ideal_time_stamp = "2023-11-05 08:42:15"


# Calculate our "ideal" grid state which we can use to calculate power quality delta
ideal_result = solve_ac_pf(sys, Ipopt.Optimizer)


# Extract the buses dictionary from the solved ideal grid
ideal_buses = ideal_result["solution"]["bus"]

# Store voltage at each bus (indexed 1 to the total number of buses)
ideal_bus_status = [ideal_buses[string(i)]["vm"] for i in 1:length(ideal_buses)]


ideal_nominal_voltage = mean(bus["vm"] for bus in values(ideal_result["solution"]["bus"]))
```

```julia
# We should see if the grid converges and find the number of islands in the grid, then add these to the global table
# Next, we must populate the subscriber's bus records with the current power quality and the status of their node

# Begin preparing global insert
using Ipopt

# For date/time
using Dates 

# For mean calculation
using Statistics

# Most recently received time stamp
time_stamp = "2023-11-05 08:42:15"


# Solve the AC Power Flow
result = solve_ac_pf(sys, Ipopt.Optimizer)

# See if grid converged
is_converged = string(result["termination_status"]) == "LOCALLY_SOLVED"

# Only caclulate islanding and PQ if the grid converges 
if is_converged

    # Get the islands
    islands = calc_connected_components(sys)
    
    num_islands = length(islands)

    # Calculate the mean voltage magnitude directly from the solution dictionary
    mean_global_voltage = mean(bus["vm"] for bus in values(result["solution"]["bus"]))

    # Calculate how far away from ideal we are, 0 will be "ideal"
    global_pq_avg = mean_global_voltage - ideal_nominal_voltage

    
    # Calculate pq at our bus
    # This value would be read from the meter instead of the file in a real scenario
    local_bus_voltage = result["solution"]["bus"][string(meter_bus_id)]["vm"]
    
    # Calculate how far away from ideal we are, 0 will be "ideal"
    local_pq = local_bus_voltage - ideal_bus_status[meter_bus_id]
 
else
    # No voltage since no convergence
    global_pq_avg = -100.0
    num_islands = 0
    local_pq = -100.0
    # If not converged, local voltage should be taken from sys instead of result
    # local_pq = sys["bus"][string(meter_bus_id)]["vm"] - ideal_bus_status[meter_bus_id]
    
end


# Convert convergence status to int for insertion
converged_int = Int(is_converged)

# BEGIN DB Interaction
# Postpone DB connection/ interaction until all of our computations have finished to minimize chance of disconnection
if USE_LOCAL_SQLITE
    println("\n--- CONNECTING TO LOCAL SQLITE ---")
    # Creates or opens a file named "test_database.sqlite" in the current directory
    conn_sq = connect_sqlite()

    # Add a global record (To represent the total grid state) if no record from last 5 minutes
    insert_global_record_sqlite(conn_sq, time_stamp, global_pq_avg, num_islands, converged_int)
    
    # Add this meter's local reading to the record table
    insert_record_sqlite(conn_sq, time_stamp, local_pq, converged_int, meter_bus_id)
    
    # SQLite closes automatically 
    println("\n--- DISCONNECTING FROM SQLITE ---")
end

if USE_AWS_POSTGRES
    println("\n--- CONNECTING TO AWS POSTGRESQL ---")
    # Connect to the AWS RDS instance via local SSH tunnel
    conn_pg = connect_pg()

    # Add a global record (To represent the total grid state) if no record from last 5 minutes
    insert_global_record_pg(conn_pg, time_stamp, global_pq_avg, num_islands, converged_int)    

    # Add this meter's local reading to the record table
    insert_record_pg(conn_pg, time_stamp, local_pq, converged_int, meter_bus_id)

    close_pg(conn_pg)
    println("\n--- DISCONNECTING FROM AWS POSTGRESQL ---")
end


println("Successfully processed and saved data for $time_stamp")
```
